#Plotting with categorical data

```python
sns.catplot(x='day',y='size',kind='strip',jitter=1,hue='sex',data=df)
#jitter is a function which control the noise of the data having same x and y value
```

This is the swarmplot and above one is stripplot ,there is not a huge difference bw them but only is distribution in data

```python
sns.swarmplot(x='day',y='tip',hue='sex',data=df)

#or sns.catplot(x='day',y='tip',kind='swarm',hue='sex',data=df)
```

#Working on heatmap

```python
df2=sns.load_dataset('flights')
df2.iloc[0:1,:]
```

```python
df3=df2.pivot_table(index='year',columns='month',values='passengers',aggfunc='mean')
```

```python
sns.heatmap(df3,cbar=True)
```

```python
sns.heatmap(df3,cbar=True,linewidth=0.2,linecolor='red',annot=True,fmt=".0f",cmap='twilight')
# You can choose any color combination you want print(plt.colormaps())
```

#clusterMap



```python
sns.clustermap(df3,cmap="winter",linewidth=0.1,linecolor='black',z_score=0,annot=True,row_cluster=False,vmin=0)
```

#About jointplot
It shows in 4 manner<br>
1.scatterplot<br>
2.Hexbinplot<br>
3.kdeplot<br>
4.Regplot

```python
sns.jointplot(x='total_bill',y='tip',kind='scatter',data=df,hue='sex')
```

```python
sns.jointplot(x='total_bill',y='tip',kind='hex',data=df)
sns.jointplot(x='total_bill',y='tip',kind='kde',data=df,hue='sex')
sns.jointplot(x='total_bill',y='tip',kind='reg',data=df)


```
